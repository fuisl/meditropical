```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Load your data
df = pd.read_csv("results/combined_metrics.csv")  # change path if needed

# If your model column is named differently, change this
model_col = "Model"

metrics = ["Faithfulness", "Answer Accuracy", "Answer Relevance", "Reasoning Recall"]

# Pastel theme + pastel palette
sns.set_theme(style="whitegrid")
palette = sns.color_palette("pastel")

fig, axes = plt.subplots(2, 2, figsize=(14, 8), sharex=True)
axes = axes.flatten()

for ax, metric in zip(axes, metrics):
    sns.barplot(
        data=df,
        x=model_col,
        y=metric,
        ax=ax,
        palette=palette,
        ci=None
    )
    ax.set_title(metric)
    ax.set_xlabel("")
    ax.set_ylabel(metric)
    ax.tick_params(axis="x", rotation=30)

# If there are extra axes (shouldn't be), hide them
for i in range(len(metrics), len(axes)):
    axes[i].set_visible(False)

plt.tight_layout()
plt.show()
```
