```python
import re
import os
```

```python
def process_figures(file_path):
    with open(file_path, 'r', encoding='utf-8') as f:
        content = f.read()

    # Extract case number from filename
    filename = os.path.basename(file_path)
    case_match = re.match(r'case_(\d+)', filename)
    if not case_match:
        print("Could not extract case number from filename")
        return
    case_num = case_match.group(1)

    # Extract all image paths in order
    image_pattern = r'!\[\]\((images/[^)]+)\)'
    image_paths = re.findall(image_pattern, content)

    print(f"Case: {case_num}, Found {len(image_paths)} images")

    # Remove image markdown and their captions
    content = re.sub(r'!\[\]\(images/[^)]+\)\s*\n\s*•[^\n]*\n?', '', content)

    # Create a mapping of Fig number to image path
    fig_to_image = {}
    for idx, img_path in enumerate(image_paths):
        fig_num = idx + 1  # Fig numbering starts at 1
        fig_to_image[fig_num] = img_path

    # Replace Fig. references with appended image paths as markdown links
    def replace_fig(match):
        fig_ref = match.group(0)  # e.g., "Fig. 11.1"

        # Extract the figure number (the part after the case number)
        fig_num_match = re.search(r'Fig\.\s*\d+\.(\d+)', fig_ref)
        if fig_num_match:
            fig_num = int(fig_num_match.group(1))
            if fig_num in fig_to_image:
                replacement = f"{fig_ref} ![]({fig_to_image[fig_num]})"
                return replacement
        return fig_ref

    # Match "Fig." followed by case.image format
    content = re.sub(r'Fig\.\s*\d+\.\d+', replace_fig, content)

    # Write back to file
    with open(file_path, 'w', encoding='utf-8') as f:
        f.write(content)

    print(f"Processed figures successfully")
```

```python
for case_report in os.listdir("/content/extracted_case_report"):
  case_count = case_report.split("---")

  os.rename(f"/content/extracted_case_report/{case_report}", f"/content/extracted_case_report/case_{case_count[0]}")
```

```python
for case_report in os.listdir("/content/extracted_case_report"):
  for auto in os.listdir(f"/content/extracted_case_report/{case_report}"):
    for filename in os.listdir(f"/content/extracted_case_report/{case_report}/{auto}"):
      if (filename != "images"):
        case_count = filename.split("---")

        os.rename(f"/content/extracted_case_report/{case_report}/{auto}/{filename}", f"/content/extracted_case_report/{case_report}/{auto}/case_{case_count[0]}.md")
```

```python
for case_report in os.listdir("/content/extracted_case_report"):
  process_figures(f"/content/extracted_case_report/{case_report}/auto/{case_report}.md")
```
